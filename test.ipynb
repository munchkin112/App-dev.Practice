```python
import urllib.request
import urllib.error
import datetime  # 예제용

# 예제 URL
target_url = "http://127.0.0.1:8000/books"

# HTTP 테스트용 사이트
# target_url = "http://httpbin.org/get"

print(f"--- URL GET 요청 시작: {target_url} ---")

try:
    # with 문을 사용하여 자동으로 리소스 정리
    with urllib.request.urlopen(target_url, timeout=10) as response:
        status = response.status

        print(f"상태 코드: {status}")

        headers = response.headers

        print(
            f"콘텐츠 타입: "
            f"{headers.get('Content-Type', '알 수 없음')}"
        )

        # 헤더에서 인코딩 정보 얻기
        # 없으면 utf-8로 가정
        encoding = headers.get_content_charset() or "utf-8"

        print(f"감지된/가정된 인코딩: {encoding}")

        # 200번대 응답인지 확인
        if 200 <= status < 300:
            # 본문을 바이트(Byte, 바이트) 형태로 읽기
            html_bytes = response.read()

            # 바이트를 문자열로 디코딩
            html_content = html_bytes.decode(encoding)

            print("\n--- 응답 내용 (일부) ---")

            # 앞 500자만 출력
            print(html_content[:500] + "...")

        else:
            print(f"오류 응답 수신: 상태 코드 {status}")


# HTTP 오류 처리
# 예: 404, 500
except urllib.error.HTTPError as e:
    print(f"HTTP 오류 발생: {e.code} {e.reason}")


# URL 관련 오류 처리
# 예: 사이트 주소 오류, 네트워크 연결 불가
except urllib.error.URLError as e:
    print(f"URL 오류 발생: {e.reason}")


# 요청 시간이 너무 오래 걸릴 경우
except TimeoutError:
    print("요청 시간 초과")


# 그 외 예상하지 못한 오류
except Exception as e:
    print(f"기타 오류 발생: {e}")
```
